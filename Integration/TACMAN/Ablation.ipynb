```python
p_homo = Path('~/link/TACMAN/homo').expanduser()


def get_homo_path(
        sp_ref,
        sp_que,
        p_homo_info
):
    p_homo_info = Path(p_homo_info)
    df_info = pd.read_csv(p_homo_info)
    index_ = df_info.query("sp_ref == '{}' & sp_que == '{}'".format(sp_ref, sp_que)).index
    assert index_.size == 1, "[get {} path]can not get speicifed and unique path\nsp_ref\tsp_que\n{}\t{}".format(
        index_.size, sp_ref, sp_que)
    res = Path(df_info.loc[index_[0], 'path'])
    if not res.is_absolute():
        res = p_homo_info.parent.joinpath(res)
    assert res.exists(), "[not exists] {}".format(res)
    return res

get_homo_path('human','mouse',p_homo.joinpath('info.csv'))\
    .__str__()[36:] # Hide my path
```

```python
gene_list = 'IFI44L', 'NCAM1', 'SCARB2', 'TNFSF4'
gene_list2 = 'Ifi44l', 'Ncam1', 'Scarb2', 'Tcfsf4'

# 3. 过滤函数：删除指定基因，只保留存在的基因
def filter_genes(adata, genes_to_remove):
    # 只保留在 adata 中实际存在的基因
    exist_genes = [g for g in genes_to_remove if g in adata.var_names]
    # 取反：保留不在删除列表里的基因
    adata = adata[:, ~adata.var_names.isin(exist_genes)].copy()
    print(f"实际存在的基因{exist_genes}")
    print(f"删除了 {len(exist_genes)} 个基因")
    return adata

# 4. 执行删除
adata_ref = filter_genes(adata_ref, human_genes) #11
adata_que = filter_genes(adata_que, mouse_genes) #9
```

```python
# ====================== 3. 检查函数 ======================
def check_genes_exist(adata, gene_list, name):
    print(f"====== 检查 {name} 数据 ======")
    still_exist = [g for g in gene_list if g in adata.var_names]
    already_gone = [g for g in gene_list if g not in adata.var_names]
    
    print(f"✅ 已成功删除的基因 ({len(already_gone)} 个):")
    print(already_gone)
    
    print(f"\n⚠️ 仍然存在的基因 ({len(still_exist)} 个):")
    print(still_exist)
    
    print("-" * 50)
    return still_exist

# ====================== 4. 开始检查 ======================
print("🔍 开始检查过滤后的 adata 文件...\n")

# 检查人类数据
human_still = check_genes_exist(adata_ref, human_genes, "人类")

# 检查小鼠数据
mouse_still = check_genes_exist(adata_que, mouse_genes, "小鼠")

# ====================== 5. 最终结论 ======================
total_still = len(human_still) + len(mouse_still)
if total_still == 0:
    print("🎉 全部删除成功！目标基因已不在数据中！")
else:
    print(f"❌ 还有 {total_still} 个基因未删除")
```

```python
import numpy as np
import scanpy as sc
from sklearn.neighbors import NearestNeighbors

# 1. 物种隔离指数
def calculate_isolation_score(adata, label_key='species', k=50, use_rep='X_umap'):
    """计算隔离指数。值越低，混合越好。"""
    # 获取嵌入和标签
    X = adata.obsm[use_rep]
    labels = adata.obs[label_key].values
    unique_labels = np.unique(labels)
    
    # 计算k近邻
    nbrs = NearestNeighbors(n_neighbors=k+1).fit(X) # 自己也是邻居
    distances, indices = nbrs.kneighbors(X)
    neighbor_indices = indices[:, 1:]  # 去掉自己
    
    # 计算每个细胞的“同类型邻居比例”
    isolation_scores = []
    for i, idxs in enumerate(neighbor_indices):
        own_label = labels[i]
        neighbor_labels = labels[idxs]
        same_label_ratio = np.mean(neighbor_labels == own_label)
        isolation_scores.append(same_label_ratio)
    
    return np.mean(isolation_scores)
```

```python
def get_mouse_homologs_from_human(
    human_genes: list,
    homology_csv: str = "homo/human_to_mouse.txt"
) -> list:
    """
    输入人类基因列表，自动输出同源小鼠基因列表
    适用于：免疫细胞活化与分化模块 / 跨物种整合分析
    
    参数：
        human_genes: 人类基因列表
        homology_csv: 人-鼠同源基因映射文件（默认：homo/human_to_mouse.txt）
    
    返回：
        同源小鼠基因列表（无匹配则为 NO_MATCH）
    """
    # 建立 人基因 → 小鼠基因 的映射字典
    gene_map = {}
    with open(homology_csv, "r", encoding="utf-8") as f:
        next(f)  # 跳过表头
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split(",")
            if len(parts) >= 2:
                human = parts[0].strip()
                mouse = parts[1].strip()
                gene_map[human] = mouse

    # 批量匹配
    matched_mouse = [gene_map.get(gene, "NO_MATCH") for gene in human_genes]
    
    return matched_mouse
```
